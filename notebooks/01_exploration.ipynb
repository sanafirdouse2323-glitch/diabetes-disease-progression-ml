# 01 — Explore the diabetes regression dataset

I start by looking at the public scikit-learn diabetes dataset and the one-year disease-progression target. This notebook uses the unstandardized source feature values returned by `load_diabetes(scaled=False)`. The dataset has 442 observations and 10 features. It is a research benchmark, not a collection of patient records or a clinical prediction service.

```python
import matplotlib.pyplot as plt
import pandas as pd

from diabetes_ml.data import load_diabetes_data

X, y = load_diabetes_data()
print(f"Observations: {len(X)}")
print(f"Features: {X.shape[1]}")
print(f"Missing feature values: {int(X.isna().sum().sum())}")
print(f"Missing target values: {int(y.isna().sum())}")
display(X.head())
```

```python
display(X.describe().T)
display(y.describe().to_frame(name="one_year_progression_measure"))
```

```python
y.hist(bins=25)
plt.title("Distribution of the continuous progression target")
plt.xlabel("Target value")
plt.ylabel("Observations")
plt.tight_layout()
plt.show()
```

```python
correlations = X.assign(target=y).corr(numeric_only=True)["target"].drop("target").sort_values()
display(correlations.to_frame("Pearson correlation with target"))
```

## What the data tells us

The loader returns the dataset's unstandardized source feature values. Any scaling needed by a model is fitted later, using only the relevant training partition or cross-validation fold. The correlations in this notebook are descriptive; they do not show that one feature causes a change in disease progression. The dataset is small and does not provide the clinical context needed to establish medical usefulness.
