# 01 — Explore the diabetes regression dataset

This notebook introduces the public, preprocessed scikit-learn diabetes dataset. Its target is a continuous measure of disease progression one year after baseline. This is a small educational benchmark, not patient-level clinical data or a clinical prediction service.

```python
import matplotlib.pyplot as plt
import pandas as pd

from diabetes_ml.data import load_diabetes_data

X, y = load_diabetes_data()
print(f"Observations: {len(X)}")
print(f"Features: {X.shape[1]}")
print(f"Missing feature values: {int(X.isna().sum().sum())}")
print(f"Missing target values: {int(y.isna().sum())}")
display(X.head())
```

```python
display(X.describe().T)
display(y.describe().to_frame(name="one_year_progression_measure"))
```

```python
y.hist(bins=25)
plt.title("Distribution of the continuous progression target")
plt.xlabel("Target value")
plt.ylabel("Observations")
plt.tight_layout()
plt.show()
```

```python
correlations = X.assign(target=y).corr(numeric_only=True)["target"].drop("target").sort_values()
display(correlations.to_frame("Pearson correlation with target"))
```

## Interpretation

The feature values are already preprocessed and standardised by the source dataset. Correlations are descriptive associations, not causal effects. The dataset contains 442 observations and 10 features; it does not provide the scale or clinical context needed to establish medical usefulness.
